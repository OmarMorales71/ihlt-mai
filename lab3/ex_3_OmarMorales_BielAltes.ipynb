## NLTK Tokenizer

````python
import nltk
nltk.download('punkt_tab')

dt_tokens = dt.loc[:, [0, 1]].applymap(nltk.word_tokenize)
dt_tokens.head(10)
dt_tokens[0][0]
````

## Compute the Jaccard similarity

````python
from nltk.metrics.distance import jaccard_distance
dt['jaccard'] = dt_tokens.apply(lambda row: jaccard_distance(set(row[0]), set(row[1])), axis=1)
dt.head(10)
````

## Compute the pearson correlation



````python
from scipy.stats import pearsonr
pearsonr(dt['gs'], dt['jaccard'])[0]
````

## Justification



We got a negative correlation of about -0.45 because the two scores measure opposite things. The ground truth score (gs) goes up when sentences are very similar, but the Jaccard distance score goes up when sentences are very different. Since high similarity means a low distance, one score goes up while the other goes down, which is why the correlation is negative. The correlation scale goes from -1 to 1. A score of -1 means a perfect opposite relationship, 0 means no relation at all, and 1 means a perfect match. Our result of -0.45 simply means they have a clear, medium-strength opposite relationship.
