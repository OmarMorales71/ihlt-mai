```python
from pathlib import Path
import tarfile

path = Path('../sts/resources/test-gold.tgz')

with tarfile.open(path) as tar:
    pairs = [line.split('\t') for line in tar.extractfile('test-gold/STS.input.SMTeuroparl.txt').read().decode('utf-8').splitlines()]
    gs = [float(line) for line in tar.extractfile('test-gold/STS.gs.SMTeuroparl.txt').read().decode('utf-8').splitlines()]

len(pairs), len(gs)
```

```python
print(f"{'#':<3} {'Sentence 1':<65} Sentence 2")
for i, (a, b) in enumerate(pairs[:5], 1):
    print(f"{i:<3} {a[:62]:<65} {b[:62]}")
```

## Jaccard distance

```python
import re
from nltk.metrics import jaccard_distance

def words(sentence):
    return set(re.findall(r'\b\w+\b', sentence.casefold()))

distances = [jaccard_distance(words(a), words(b)) for a, b in pairs]
len(distances)
```

```python
print(f"{'Pair':<6} {'Gold':<8} Jaccard distance")
for i in range(10):
    print(f"{i + 1:<6} {gs[i]:<8.3f} {distances[i]:.3f}")
```

## Pearson correlation

```python
from statistics import correlation

r = correlation(gs, distances)
r
```

The correlation is negative because the gold scores measure similarity, while Jaccard distance measures difference. Word overlap also misses paraphrases that use different words.
