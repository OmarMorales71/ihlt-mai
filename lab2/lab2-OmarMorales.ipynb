## NLTK Tokenizer

````python
import nltk
nltk.download('punkt_tab')

dt_tokens = dt.loc[:, [0, 1]].applymap(nltk.word_tokenize)
dt_tokens.head(10)
````

## Compute the Jaccard similarity

````python
from nltk.metrics.distance import jaccard_distance
dt['jaccard'] = dt_tokens.apply(lambda row: jaccard_distance(set(str(row[0]).split()), set(str(row[1]).split())), axis=1)
dt.head(10)
````

## Compute the pearson correlation



````python
from scipy.stats import pearsonr
pearsonr(dt['gs'], dt['jaccard'])[0]
````
